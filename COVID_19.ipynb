```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Covid Data.csv")

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)
```

```python
print(df.head())
```

```python
print(df.tail())
```

```python
print("Column names:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing value percentage:")
print(missing_percentage.sort_values(ascending=False))
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
print(df.describe())
```

```python
for column in df.columns:
    print("\n", column)
    print(df[column].value_counts().head(10))
```

```python
df["DEATH"] = df["DATE_DIED"].apply(
    lambda x: 0 if x == "9999-99-99" else 1
)

print(df["DEATH"].value_counts())
```

```python
print(df["DATE_DIED"].value_counts().head(10))
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["AGE"], bins=30, kde=True)

plt.title("Age Distribution of COVID-19 Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.countplot(x="SEX", data=df)

plt.title("Gender Distribution of COVID-19 Patients")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.countplot(x="PNEUMONIA", data=df)

plt.title("Pneumonia Distribution")
plt.xlabel("Pneumonia")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.countplot(x="DIABETES", data=df)

plt.title("Diabetes Distribution Among Patients")
plt.xlabel("Diabetes")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.countplot(x="HIPERTENSION", data=df)

plt.title("Hypertension Distribution Among Patients")
plt.xlabel("Hypertension")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.countplot(x="PATIENT_TYPE", data=df)

plt.title("Patient Type Distribution")
plt.xlabel("Patient Type")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(x="CLASIFFICATION_FINAL", data=df)

plt.title("COVID-19 Classification Distribution")
plt.xlabel("Classification")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(x="DEATH", y="AGE", data=df)

plt.title("Age Distribution by Death Status")
plt.xlabel("Death Status")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="DIABETES", hue="DEATH", data=df)

plt.title("Diabetes vs Death")
plt.xlabel("Diabetes")
plt.ylabel("Number of Patients")
plt.legend(title="Death")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="PNEUMONIA", hue="DEATH", data=df)

plt.title("Pneumonia vs Death")
plt.xlabel("Pneumonia")
plt.ylabel("Number of Patients")
plt.legend(title="Death")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="HIPERTENSION", hue="DEATH", data=df)

plt.title("Hypertension vs Death")
plt.xlabel("Hypertension")
plt.ylabel("Number of Patients")
plt.legend(title="Death")

plt.show()
```

```python
plt.figure(figsize=(14, 10))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=False,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of COVID-19 Dataset")
plt.show()
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```
